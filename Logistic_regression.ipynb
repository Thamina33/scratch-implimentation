```python
import numpy as np
```

```python
class LogisticRegression:
    def __init__(self, learning_rate = 0.01, n_iter = 1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter

    def _sigmoid(self, z):
        return (1/(1+np.exp(-z)))

    def fit(self , X, y):
        m,n = X.shape

        self.bias = 0
        self.weights = np.zeros(n)

        for i in range(self.n_iter):
            z = self.bias + X @ self.weights
            y_pred = self._sigmoid(z)

            db = (1/m) * (y_pred-y)
            dw = (1/m) * X.T @ (y_pred-y)

            self.bias -= self.lr * db
            self.weights -= self.lr*dw

    def get_probalities(self , X):
        z = self.bias + X @ self.weights
        return self._sigmoid(z)
    
    def predict(self, X , Thresold = 0.5):
        probabilites = self.get_probalities(X)
        y_pred_bool = probabilites >= Thresold

        return y_pred_bool.astype(int)

    
```

```python
x = np.array([[1,2] , [2,3] , [3,4] , [4,5]])
y = np.array([1, 0, 1, 0])

model = LogisticRegression()
model.fit(x,y)

y_pred = model.predict(x)
print(y_pred)
print("Probabilities:", model.get_probalities(x))
```
